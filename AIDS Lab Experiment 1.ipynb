## Assignment 1

```python
import numpy as np
```

```python
# function for addition
def add(a, b):
  return a + b
```

```python
# function for subtraction
def subtract(a, b):
  return a-b
```

```python
# function for multiplication
def multiply(a, b):
  return a*b
```

```python
# function for division
def divide(a, b):
  return a / b
```

```python
# function for dot product
def dot_product(a, b):
  return np.dot(a, b)
```

```python
# declaring two arrays
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
```

```python
print("Matrix A:")
print(A)
print("Matrix B")
print(B)
print("Matrix Addition")
print(add(A, B))
print("Matrix Subtraction")
print(subtract(A, B))
print("Matrix Multiplication")
print(multiply(A, B))
print("Matrix Dot Product")
print(dot_product(A, B))
print("Matrix Divide")
print(divide(A, B))
```

## Assignment 2

```python
C = np.array([[1, 2],
 [3, 4]])
```

```python
# Determinant
print("Matrix C:\n", C)
print("Determinant:\n", np.linalg.det(C))
```

```python
# Inverse of Matrix
print("Inverse of Matrix: ")
np.linalg.inv(C)
```

```python
# Transpose of Matrix
print("Transpose of Matrix (using linalg):")
np.linalg.matrix_transpose(C)
```

```python
# Transpose of Matrix (Direct Method)
print("Transpose of Matrix (Direct Method):")
np.transpose(C)
```
