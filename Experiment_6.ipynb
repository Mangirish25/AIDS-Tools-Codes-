```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 34, 42],
    "Salary": [25000, 30000, 35000, 40000]
}

df = pd.DataFrame(data)
df
```

```python
# returns True where value is missing
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
(df.isnull().sum() / len(df)) * 100
```

```python
# returns False where value is missing
df.notnull()
```

```python
mean = df["Age"].mean()
df["Age"].fillna(mean, inplace=True)
df
```

```python
df.replace(25, 26, inplace=True)
```

```python
df
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 34, 42],
    "Salary": [25000, np.nan, 35000, 40000]
}

df = pd.DataFrame(data)

changed = df.interpolate(method="linear", inplace=False)
changed
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", None],
    "Age": [25, np.nan, 34, 42],
    "Salary": [25000, np.nan, 35000, 40000]
}

df = pd.DataFrame(data)

changed = df.dropna(inplace=False)
changed
```
