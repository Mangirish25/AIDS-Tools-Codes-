```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# training data
x = np.array([1, 2, 3, 5])
y = np.array([20, 40, 50, 60])
```

```python
# calculate the slope and intercept

n = len(x)

m = (n * np.sum(x*y) - np.sum(x) * np.sum(y)) / (n * np.sum(x ** 2) - (np.sum(x)) ** 2)
c = (np.sum(y) - m * np.sum(x)) / n
```

```python
print("Model Equation y = " , m , "x + ", c)
```

```python
y_pred = m*x + c

x_new = 5
y_new = m*x_new + c
print("Prediction for x = ", x_new, " is ", y_new)
```

```python
plt.scatter(x, y, color = "blue", label = "Training Data")
plt.plot(x, y_pred, color = "red", label = "Regression line")
plt.scatter(x_new, y_new, color = "green", s = 100, label = "Predicted Point")
plt.xlabel("Study hours (x)")
plt.ylabel("Marks (Y)")
plt.title("Linear Regression Visualization")
plt.legend()
plt.grid()
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
X = np.array([
    [1000, 2],
    [1500, 3],
    [1800, 3],
    [2000, 4],
    [2200, 4]
])

y = np.array([200, 250, 270, 300, 320])
```

```python
ones = np.ones((X.shape[0], 1))
X_b = np.hstack((ones, X))
```

```python
#  Calculate coefficients
beta = np.linalg.inv(X_b. T @ X_b) @ X_b.T @ y
```

```python
print("Intercept: ", beta[0])
print("Size coefficient: ", beta[1])
print("Bedroom coefficient: ", beta[2])
```

```python
size_range = np.linspace(1000, 2200, 10)
bedroom_range = np.linspace(2, 4, 10)

size_grid, bedroom_grid = np.meshgrid(size_range, bedroom_range)

price_grid = beta[0] + beta[1] * size_grid + beta[2]*bedroom_grid
```

```python
new_house = np.array([1, 2100, 3])
predicted_price = new_house @ beta
print("Predicted Price: ", predicted_price)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection = "3d")

ax.scatter(X[: ,0], X[: , 1], y, color = 'blue', label = "Training Data")

ax.plot_surface(size_grid, bedroom_grid, price_grid, alpha = 0.5)

ax.scatter(2100, 3, predicted_price, color = "green", s = 100, label = "Predicted Price")

ax.set_ylabel("Bedrooms")
ax.set_xlabel("House Size")
ax.set_zlabel("Price")

ax.set_title("Multiple Linear Regression (3D Visualization)")

ax.legend()
plt.show()
```
